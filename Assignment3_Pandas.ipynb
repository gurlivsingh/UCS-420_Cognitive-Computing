### Q4. Read a CSV File (Iris Dataset)

### Q5. Delete Row 4 and Column 3 from Iris Dataset

### Q6. Employees Dataset Analysis

```python
import pandas as pd
import io

employees_data = """Employee_ID,Name,Department,Age,Salary,Years_of_Experience,Joining_Date,Gender,Bonus,Rating
101,Alice,HR,29,50000,4,2020-03-15,Female,5000,4.5
102,Bob,IT,34,70000,8,2017-07-19,Male,7000,4.0
103,Charlie,IT,41,65000,10,2013-06-01,Male,6000,3.8
104,Diana,Marketing,28,55000,3,2021-02-10,Female,4500,4.7
105,Edward,Sales,38,60000,12,2010-11-25,Male,5000,3.5"""

emp_df = pd.read_csv(io.StringIO(employees_data))

# a) Shape
print("a) Shape:", emp_df.shape)

# b) Summary (dtypes and non-null counts)
print("\nb) DataFrame Info:")
print(emp_df.info())

# c) Descriptive statistics
print("\nc) Descriptive Statistics:")
print(emp_df.describe())

# d) First 5 and last 3 rows
print("\nd) First 5 rows:")
print(emp_df.head(5))
print("\nLast 3 rows:")
print(emp_df.tail(3))

# e) Statistics
print("\ne) Statistics:")
print(f"  i. Average Salary: {emp_df['Salary'].mean()}")
print(f"  ii. Total Bonus: {emp_df['Bonus'].sum()}")
print(f"  iii. Youngest Employee Age: {emp_df['Age'].min()}")
print(f"  iv. Highest Rating: {emp_df['Rating'].max()}")

# f) Sort by Salary descending
print("\nf) Sorted by Salary (descending):")
print(emp_df.sort_values(by='Salary', ascending=False))

# g) Add Performance Category column
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp_df["Performance"] = emp_df["Rating"].apply(categorize)
print("\ng) With Performance Category:")
print(emp_df[["Name", "Rating", "Performance"]])

# h) Missing values
print("\nh) Missing Values:")
print(emp_df.isnull().sum())

# i) Rename Employee_ID to ID
emp_df.rename(columns={"Employee_ID": "ID"}, inplace=True)
print("\ni) Renamed column. Columns now:", list(emp_df.columns))

# j) Filter employees
print("\nj) Employees with > 5 years experience:")
print(emp_df[emp_df["Years_of_Experience"] > 5][["Name", "Years_of_Experience"]])
print("\nEmployees in IT Department:")
print(emp_df[emp_df["Department"] == "IT"][["Name", "Department"]])

# k) Add Tax column
emp_df["Tax"] = emp_df["Salary"] * 0.10
print("\nk) With Tax Column (10% of Salary):")
print(emp_df[["Name", "Salary", "Tax"]])

# l) Save to CSV
emp_df.to_csv("employees_modified.csv", index=False)
print("\nl) Saved modified DataFrame to employees_modified.csv")
```
