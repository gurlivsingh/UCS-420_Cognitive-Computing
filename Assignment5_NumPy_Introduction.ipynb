### Q1. Create a NumPy 1D array with 5 elements and perform basic operations

```python
arr = np.array([5, 10, 15, 20, 25])
print("Original 1D Array:", arr)

# a) Addition of 2 in all elements
print("\na) Addition of 2:", arr + 2)

# b) Multiply 3 with all elements
print("b) Multiply by 3:", arr * 3)

# c) Divide every element by 2
print("c) Divide by 2:", arr / 2)
```

### Q2. Basic NumPy Array Operations

```python
# a) Reverse the NumPy array
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr)
print("a) Reversed array:", arr[::-1])

# b) Find most frequent value and their indices
print("\nb) Most Frequent Value:")

x = np.array([1,2,3,4,5,1,2,1,1,1])
print("   i. x =", x)
values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
most_freq = values[counts == max_count]
indices = [np.where(x == v)[0].tolist() for v in most_freq]
print(f"      Most frequent value: {most_freq}, count: {max_count}")
print(f"      Indices: {indices}")

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
print("\n   ii. y =", y)
values_y, counts_y = np.unique(y, return_counts=True)
max_count_y = counts_y.max()
most_freq_y = values_y[counts_y == max_count_y]
indices_y = [np.where(y == v)[0].tolist() for v in most_freq_y]
print(f"       Most frequent value: {most_freq_y}, count: {max_count_y}")
print(f"       Indices: {indices_y}")
```

### Q4. linspace() Array - Dimensions, Shape, Dtype, Transpose

```python
# Create 1D array with 25 evenly spaced numbers from 10 to 100
Gurlivsingh = np.linspace(10, 100, 25)
print("Array 'Gurlivsingh' (25 evenly spaced numbers from 10 to 100):")
print(Gurlivsingh)

print("\nDimensions (ndim):", Gurlivsingh.ndim)
print("Shape:", Gurlivsingh.shape)
print("Total elements (size):", Gurlivsingh.size)
print("Data type (dtype):", Gurlivsingh.dtype)
print("Total bytes:", Gurlivsingh.nbytes)

# Find transpose using reshape()
print("\nTranspose using reshape():")
transposed_reshape = Gurlivsingh.reshape(25, 1)
print(transposed_reshape[:5], "... (shape:", transposed_reshape.shape, ")")

# Transpose using .T attribute
print("\nTranspose using .T attribute (on reshaped 2D):")
arr_2d = Gurlivsingh.reshape(5, 5)
print("Reshaped to 5x5:")
print(arr_2d)
print("\nTransposed using .T:")
print(arr_2d.T)
print("\nNote: .T on a 1D array has no effect (shape stays the same)")
```

### Q5. 2D Array - Mean, Median, Max, Min, Unique, Reshape, Resize

```python
# Create 2D array 3x4
ucs420_Gurlivsingh = np.array([[10, 20, 30, 40],
                                [50, 60, 70, 80],
                                [90, 15, 20, 35]])

print("ucs420_Gurlivsingh (3 rows x 4 columns):")
print(ucs420_Gurlivsingh)

print("\nMean:", np.mean(ucs420_Gurlivsingh))
print("Median:", np.median(ucs420_Gurlivsingh))
print("Max:", np.max(ucs420_Gurlivsingh))
print("Min:", np.min(ucs420_Gurlivsingh))
print("Unique elements:", np.unique(ucs420_Gurlivsingh))

# Reshape to 4 rows x 3 columns
reshaped_ucs420_Gurlivsingh = ucs420_Gurlivsingh.reshape(4, 3)
print("\nreshaped_ucs420_Gurlivsingh (4 rows x 3 columns):")
print(reshaped_ucs420_Gurlivsingh)

# Resize to 2 rows x 3 columns
resized_ucs420_Gurlivsingh = np.resize(ucs420_Gurlivsingh, (2, 3))
print("\nresized_ucs420_Gurlivsingh (2 rows x 3 columns):")
print(resized_ucs420_Gurlivsingh)
```
